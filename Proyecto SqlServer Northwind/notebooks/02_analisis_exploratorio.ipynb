```python
import pandas as pd
import numpy as np

# Cargar datos
df = pd.read_csv('../data/northwind_processed.csv')
df['OrderDate'] = pd.to_datetime(df['OrderDate'])

# 1. Limpieza e Inspección Básica
print("--- Información de Columnas y Tipos de Datos ---")
print(df.info())

print("\n--- Conteo de Valores Nulos ---")
print(df.isnull().sum())

# Tratamiento de nulos en Región
df['CustomerRegion'] = df['CustomerRegion'].fillna('Sin Región')

# 2. Análisis Estadístico Descriptivo
print("\n--- Resumen Estadístico de Variables Numéricas ---")
print(df[['UnitPrice', 'Quantity', 'Discount', 'TotalSales']].describe())

# 3. Detección de Valores Atípicos (Outliers) en Precios/Ventas
q1 = df['TotalSales'].quantile(0.25)
q3 = df['TotalSales'].quantile(0.75)
iqr = q3 - q1
limite_superior = q3 + 1.5 * iqr

outliers = df[df['TotalSales'] > limite_superior]
print(f"\nCantidad de transacciones atípicas (altos volúmenes): {len(outliers)}")

# 4. Matriz de Correlación
correlacion = df[['UnitPrice', 'Quantity', 'Discount', 'TotalSales']].corr()
print("\n--- Matriz de Correlación ---")
print(correlacion)
```
